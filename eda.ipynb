# 📊 Data Dictionary — Marketing Campaign Dataset

## 🎯 Objetivo do Dataset

Este dataset contém informações de clientes e seu comportamento de compra, com o objetivo de analisar e prever a resposta a campanhas de marketing.

---

# 🧾 👤 Identificação

### ID

* Identificador único do cliente
* Não possui significado analítico direto
* Pode ser usado como chave primária

---

# 🎂 👥 Dados Demográficos

### Year_Birth

* Ano de nascimento do cliente
* Pode ser transformado em idade
* Útil para segmentação demográfica

### Education

* Nível educacional do cliente
* Exemplos: Graduation, PhD, Master
* Pode indicar poder aquisitivo e perfil de consumo

### Marital_Status

* Estado civil do cliente
* Exemplos: Single, Married, Divorced
* Pode influenciar comportamento de compra

### Income

* Renda anual do cliente
* Uma das variáveis mais importantes
* Pode conter valores faltantes e outliers

### Kidhome

* Número de crianças na casa
* Pode impactar padrão de consumo

### Teenhome

* Número de adolescentes na casa
* Também influencia comportamento de compra

---

# 🛒 💰 Comportamento de Consumo

### MntWines

* Valor gasto em vinhos

### MntFruits

* Valor gasto em frutas

### MntMeatProducts

* Valor gasto em carne

### MntFishProducts

* Valor gasto em peixe

### MntSweetProducts

* Valor gasto em doces

### MntGoldProds

* Valor gasto em produtos premium ("gold")

👉 Todas essas variáveis representam o consumo por categoria

---

# 🛍️ 📦 Comportamento de Compra

### NumDealsPurchases

* Número de compras com desconto

### NumWebPurchases

* Número de compras realizadas via site

### NumCatalogPurchases

* Número de compras via catálogo

### NumStorePurchases

* Número de compras em loja física

### NumWebVisitsMonth

* Número de visitas ao site no último mês

---

# ⏱️ 📅 Engajamento Temporal

### Recency

* Número de dias desde a última compra
* Quanto menor, mais recente o cliente

### Dt_Customer

* Data de entrada do cliente na base
* Pode ser usada para calcular tempo de relacionamento

---

# 📢 📈 Campanhas de Marketing

### AcceptedCmp1

### AcceptedCmp2

### AcceptedCmp3

### AcceptedCmp4

### AcceptedCmp5

* Indicam se o cliente aceitou campanhas anteriores
* 1 = aceitou, 0 = não aceitou
* Muito importantes para prever comportamento futuro

---

# 🎯 🎯 Variável Alvo

### Response

* Indica se o cliente respondeu à última campanha
* 1 = respondeu
* 0 = não respondeu

👉 Essa é a variável target para modelos de classificação

---

# ⚠️ ⚠️ Reclamações

### Complain

* Indica se o cliente fez reclamação nos últimos 2 anos
* 1 = sim
* 0 = não

---

# 🧠 💡 Possíveis Features Derivadas

Você pode criar novas variáveis:

* Age → a partir de Year_Birth
* Total_Spending → soma dos gastos
* Total_Purchases → soma das compras
* Customer_Tenure → tempo como cliente
* Accepted_Total → total de campanhas aceitas

---

* Dados demográficos
* Comportamento de compra
* Histórico de campanhas
* Engajamento do cliente

👉 Permitindo análises completas de marketing e modelagem preditiva.
